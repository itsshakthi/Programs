```python
#leetcode 2149
nums=[3,1,-2,-5,2,-4]
i=0
j=0
n = len(nums)
p=[]
while i<n and j<n:
    if nums[i]>0 and nums[j]<0:
        p.extend([nums[i],nums[j]])
        i+=1
        j+=1
    elif nums[i]>0:
        j+=1
    else:
        i+=1
print(p)
```

```python
#leetcode 2108
w = ["abc","car","ada","racecar","cool"]
p=None
for i in w:
    
    left=0
    right=len(i)-1
    check = True
    while left < right:
        if i[left] != i[right]:
            check = False
            break
        left += 1
        right -= 1  
    if check:
        p=i
        break
    else:
        print('')
print(p)
```

```python
#leetcode 3149
l=[7,8,3,4,15,13,4,1]
avg=[]
p=[]
while l:
    if not l:
        exit()
    
    else:
        m=max(l)
        n=min(l)

        s=m+n
        avg.append(s/2)
        l.remove(m)
        l.remove(n)
        
print(avg)
for ans in avg:
    if ans.is_integer():
        p.append(int(ans))
    else:
        p.append(ans)

print(min(p))
```
