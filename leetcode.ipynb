```python
#leetcode 3149
l=[7,8,3,4,15,13,4,1]
avg=[]
p=[]
while l:
    if not l:
        exit()
    
    else:
        m=max(l)
        n=min(l)

        s=m+n
        avg.append(s/2)
        l.remove(m)
        l.remove(n)
        
print(avg)
for ans in avg:
    if ans.is_integer():
        p.append(int(ans))
    else:
        p.append(ans)

print(min(p))
```

```python
#leetcode 682
ops = ["5","-2","4","C","D","9","+","+"]
r=[]
for i in ops:
    if i.lstrip('-').isdigit():
        r.append(int(i))
    elif i=='C':
        r.pop()
    elif i=='D':
        r.append((int(r[-1])*2))
    elif i=='+':
        r.append((int(r[-1])+int(r[-2])))
if r:
    print(sum(r))
else:
    print(0)
```

```python
#leetcode 20

s='{}([])'
l=[]
q='({['
for i in s:
    if i in q:
        l.append(i)
    elif i==')' and l[-1]=='(':
        l.pop()
    elif i==']' and l[-1]=='[':
        l.pop()
    elif i=='}' and l[-1]=='{':
        l.pop()
    else:
        print('False')

if l:
    print('flase')
else:
    print('true')
```

```python
#leetcode 349

nums1 = [4,9,5]
nums2 = [9,4,9,8,4]
nums1.sort()
nums2.sort()
i=0
j=0
x=[]
while i<len(nums1) and j<len(nums2):
    if nums1[i]==nums2[j]:
        if nums1[i] not in x:
            x.append(nums1[i])
            i+=1
            j+=1
        else:
            i+=1
            j+=1
    elif nums1[i]<nums2[j]:
        i+=1
    else:
        j+=1
print(x)
```
