```python
#leetcode 15
nums = [-1,0,1,2,0,-1,-2]
nums.sort()
l = []
for i in range(len(nums) - 2):
    if i > 0 and nums[i] == nums[i - 1]:
        continue
    left = i + 1
    right = len(nums) - 1
    while left < right:
        z = nums[i] + nums[left] + nums[right]
        if z == 0:
            l.append([nums[i], nums[left], nums[right]])
            while left < right and nums[left] == nums[left + 1]:
                left += 1
            while left < right and nums[right] == nums[right - 1]:
                right -= 1
            left += 1
            right -= 1
        elif z < 0:
            left += 1
        else:
            right -= 1
print(l)
```
