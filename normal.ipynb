```python
from itertools import permutations

def is_valid_chain(arr):
    for i in range(len(arr) - 1):
        if str(arr[i])[-1] != str(arr[i + 1])[0]:
            return False
    return True

def brute_force_chain(arr):
    for perm in permutations(arr):
        if is_valid_chain(perm):
            return list(perm)
    return None


# Example
print(brute_force_chain([451, 348, 123, 154]))
```

```python
s =["2","1","+","3","*"]
l=[]
x=0
for i in s:
    
    if i.lstrip('-').isdigit():
        l.append(int(i))

    elif i=='+':
        x=int(l[-2])+int(l[-1])
        l.pop()
        l.pop()
        l.append(x)
    elif i=='-':
        x=int(l[-2])-int(l[-1])
        l.pop()
        l.pop()
        l.append(x)
    elif i=='*':
        x=int(l[-2])*int(l[-1])
        l.pop()
        l.pop()
        l.append(x)
    elif i=='/':
        x=int(l[-2]/l[-1])
        l.pop()
        l.pop()
        l.append(x)
print(*l)
```

```python
h = [4,2,0,3,2,5]
l=[]
r=[]
z=h[0]
x=h[-1]
t=0
for i in range(len(h)):
    if h[i]>z:
        l.append(h[i])
        z=h[i]
    else:
        l.append(z)
print(l)
for i in range(len(h)-1,-1,-1):
    if h[i]>x:
        r.append(h[i])
        x=h[i]
    else:
        r.append(x)
r=r[::-1]
print(r)
for i in range(len(h)):
    t+=min(l[i],r[i])-h[i]
print(t)
```
